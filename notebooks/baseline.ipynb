```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import RANSACRegressor, LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from lightgbm import LGBMRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.linear_model import HuberRegressor
from sklearn.pipeline import make_pipeline
from sklearn.metrics import root_mean_squared_error
```

```python
import sys
import os

sys.path.append(os.path.abspath('..'))

import importlib
import preprocessing

importlib.reload(preprocessing)

from preprocessing import prepare_features
X, y, X_test, test_ids = prepare_features()
```

```python
n_spl = 5
kf = KFold(n_splits = n_spl, shuffle = True, random_state = 42)
```

```python
#Linear regression
fold_scores = {
    'Linear Regression': [],
    'Ridge (L2)': [],
    'Lasso (L1)': [],
    'ElasticNet (L1 + L2)': []
}

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
   
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    models = {
        'Linear Regression': LinearRegression(),
        'Ridge (L2)': Ridge(alpha=0.01),
        'Lasso (L1)': Lasso(alpha=0.01),
        'ElasticNet (L1 + L2)': ElasticNet(alpha=0.01, l1_ratio = 0.5)
    }
    
    for name, model in models.items():
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_val_scaled)
        rmse = root_mean_squared_error(y_val, y_pred)
        fold_scores[name].append(rmse)
        print(f'{name:<22} RMSE: {rmse:.2f}')

for name, scores in fold_scores.items():   
    print(f'{name:<22} Mean RMSE: {np.mean(scores):.2f}')
```

```python
#Kneighbors
n_neighbors_list = [3, 5, 7, 9, 11, 13, 15]
weights_list = ['uniform', 'distance']
metric_lists = ['euclidean', 'manhattan']
best_rmse = float('inf')
best_params = {}
for n in n_neighbors_list:
    for w in weights_list:
        for m in metric_lists:
            fold_rmse = []
            for train_idx, val_idx in kf.split(X, y):
                X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
                X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
                scaler = StandardScaler()
                X_train_scaled = scaler.fit_transform(X_train)
                X_val_scaled = scaler.transform(X_val)
    
                model = KNeighborsRegressor(
                    n_neighbors = n,
                    weights = w,
                    metric = m
                )
                model.fit(X_train_scaled, y_train)
                y_pred = model.predict(X_val_scaled)
                rmse = root_mean_squared_error(y_val, y_pred)
                fold_rmse.append(rmse)
            mean_rmse = np.mean(fold_rmse)
            print(f'k = {n:<2} | weights = {w:<8} |' f'metric = {m:<9} | RMSE = {mean_rmse:.2f}') 
            if mean_rmse < best_rmse:
                best_rmse = mean_rmse
                best_params = {
                    'n_neighbors': n,
                    'weights': w,
                    'metric': m
                }
print(f'Лучший результат: {best_rmse:.2f}')
print(f'Параметры: {best_params}')
```

```python
#Decision Tree
fold_rmse = []
for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model = DecisionTreeRegressor()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Средний RMSE: {np.mean(fold_rmse):.2f}')
```

```python
#Random forest
fold_rmse = []
best_rmse = float('inf')
best_param = 0
for n_est in range(100, 300, 30): 
    fold_rmse = []
    for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
        X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
        X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
        
        model = RandomForestRegressor(n_estimators = n_est)
        model.fit(X_train, y_train)
        y_pred = model.predict(X_val)
        rmse = root_mean_squared_error(y_val, y_pred)
        fold_rmse.append(rmse)
    mean_rmse = np.mean(fold_rmse)
    print(f'"{n_est:<2}" -> Mean RMSE = {mean_rmse:.4f}')
    
    if mean_rmse < best_rmse:
        best_rmse = mean_rmse
        best_param = n_est 
print(f'Параметр: {best_param}')
```

```python
#XGBoost
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model = XGBRegressor(
        n_estimators = 200,
        learning_rate = 0.05,
        max_depth = 4,
        random_state = 42
    )
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
#CatBoost
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model = CatBoostRegressor(
        iterations = 200,
        learning_rate = 0.05,
        depth = 4,
        verbose = 0,
        random_state = 42
    )
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
#LGBMBoost
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    model = LGBMRegressor(
        n_estimators = 200,
        learning_rate = 0.05,
        num_leaves = 15,
        max_depth = 4,
        random_state = 42,
        verbose = -1
    )
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
#MLP
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    model = MLPRegressor(
        hidden_layer_sizes = (32, 16),
        activation = 'relu',
        max_iter = 1000,
        random_state = 42
    )
    
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_val_scaled)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
#Robust regression
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    model = HuberRegressor(
        max_iter = 1000
    )
    
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_val_scaled)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
#Ransac regression
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    model = RANSACRegressor(
        estimator = LinearRegression(),
        min_samples = 0.5,
        max_trials = 200,
        random_state = 42
    )
    
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_val_scaled)
    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)
    print(f'RMSE: {rmse:.4f}')
 
print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```

```python
# Polynomial Regression
fold_rmse = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X, y), start=1):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]

    model = make_pipeline(
        PolynomialFeatures(degree=2, include_bias=False),
        StandardScaler(),
        LinearRegression()
    )

    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)

    rmse = root_mean_squared_error(y_val, y_pred)
    fold_rmse.append(rmse)

    print(f'Fold {fold}: RMSE = {rmse:.4f}')

print(f'Mean RMSE: {np.mean(fold_rmse):.4f}')
```
