```python
import numpy as np
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RANSACRegressor, LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from catboost import CatBoostRegressor
from lightgbm import LGBMRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.linear_model import HuberRegressor
from sklearn.metrics import mean_squared_log_error

def rmsle(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    y_pred = np.clip(y_pred, 0, None)
    return np.sqrt(mean_squared_log_error(y_true, y_pred))
```

```python
n_spl = 5
kf = KFold(n_splits = n_spl, shuffle = True, random_state = 42)
```

```python
#Linear regression
fold_scores = {
    'Linear Regression': [],
    'Ridge (L2)': [],
    'Lasso (L1)': [],
    'ElasticNet (L1 + L2)': []
}

for fold, (train_idx, val_idx) in enumerate (kf.split(X, y)):
    X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
    X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
   
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)
    
    models = {
        'Linear Regression': LinearRegression(),
        'Ridge (L2)': Ridge(alpha=0.01),
        'Lasso (L1)': Lasso(alpha=0.01),
        'ElasticNet (L1 + L2)': ElasticNet(alpha=0.01, l1_ratio = 0.5)
    }
    
    for name, model in models.items():
        model.fit(X_train_scaled, y_train)
        y_pred = model.predict(X_val_scaled)
        score = rmsle(y_val, y_pred)
        fold_scores[name].append(score)
        print(f'{name:<22} RMSLE: {score:.2f}')

for name, scores in fold_scores.items():   
    print(f'{name:<22} Mean RMSLE: {np.mean(scores):.2f}')
```

```python
#Kneighbors
n_neighbors_list = [3, 5, 7, 9, 11, 13, 15]
weights_list = ['uniform', 'distance']
metric_lists = ['euclidean', 'manhattan']
best_rmsle = float('inf')
best_params = {}
for n in n_neighbors_list:
    for w in weights_list:
        for m in metric_lists:
            fold_rmsle = []
            for train_idx, val_idx in kf.split(X, y):
                X_train, y_train = X.iloc[train_idx], y.iloc[train_idx]
                X_val, y_val = X.iloc[val_idx], y.iloc[val_idx]
    
                scaler = StandardScaler()
                X_train_scaled = scaler.fit_transform(X_train)
                X_val_scaled = scaler.transform(X_val)
    
                model = KNeighborsRegressor(
                    n_neighbors = n,
                    weights = w,
                    metric = m
                )
                model.fit(X_train_scaled, y_train)
                y_pred = model.predict(X_val_scaled)
                score = rmsle(y_val, y_pred)
                fold_rmsle.append(score)
            mean_rmsle = np.mean(fold_rmsle)
            print(f'k = {n:<2} | weights = {w:<8} |' f'metric = {m:<9} | RMSLE = {mean_rmsle:.2f}') 
            if mean_rmsle < best_rmsle:
                best_rmsle = mean_rmsle
                best_params = {
                    'n_neighbors': n,
                    'weights': w,
                    'metric': m
                }
print(f'Лучший результат: {best_rmsle:.2f}')
print(f'Параметры: {best_params}')
```

```python
#DNN
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from preprocessing import prepare_features
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import StandardScaler

np.random.seed(42)
torch.manual_seed(42)

X, y, X_test, test_ids = prepare_features()

X_train = X.copy()
y_train = y.copy()
X_test = X_test[X_train.columns]

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train).astype(np.float32)
X_test_scaled = scaler.transform(X_test).astype(np.float32)

y_log = np.log1p(y_train.to_numpy(dtype=np.float32))

X_tensor = torch.tensor(X_train_scaled, dtype=torch.float32)
y_tensor = torch.tensor(y_log, dtype=torch.float32).unsqueeze(1)
X_test_tensor = torch.tensor(X_test_scaled, dtype=torch.float32)

class PoliksDNN(nn.Module):
    def __init__(self, input_dim, hidden_size, activat, drop_rate, batchnorm):
        super().__init__()

        layers = []
        in_dim = input_dim

        for h_size in hidden_size:
            layers.append(nn.Linear(in_dim, h_size))

            if batchnorm:
                layers.append(nn.BatchNorm1d(h_size))

            layers.append(activat)

            if drop_rate > 0:
                layers.append(nn.Dropout(drop_rate))

            in_dim = h_size

        layers.append(nn.Linear(in_dim, 1))

        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)

hidden_size = [128, 64, 32, 16]
drop_rate = 0.2
batch_size = 64
epochs = 200
lr = 0.001

dataset = TensorDataset(X_tensor, y_tensor)
loader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=True
)

model = PoliksDNN(
    input_dim=X_train.shape[1],
    hidden_size=hidden_size,
    activat=nn.GELU(),
    drop_rate=drop_rate,
    batchnorm=True
)

loss_func = nn.MSELoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=lr,
    weight_decay=1e-5
)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=epochs
)

for epoch in range(epochs):
    model.train()
    epoch_loss = 0.0

    for x_batch, y_batch in loader:
        optimizer.zero_grad()

        output = model(x_batch)
        loss = loss_func(output, y_batch)

        loss.backward()
        optimizer.step()

        epoch_loss += loss.item() * len(x_batch)

    scheduler.step()

    if (epoch + 1) % 20 == 0:
        epoch_loss /= len(dataset)
       
model.eval()

with torch.no_grad():
    log_predictions = model(X_test_tensor).numpy().flatten()
predictions = np.expm1(log_predictions)
predictions = np.clip(predictions, 0, None)

submission = pd.DataFrame({
    'Id': test_ids.to_numpy(),
    'SalePrice': predictions
})

submission.to_csv('submission_dnn.csv', index=False)

```

```python
#GridSearchCV
import warnings
import numpy as np
import pandas as pd
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import (
    LinearRegression, Ridge, Lasso, ElasticNet, HuberRegressor
)
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_log_error, make_scorer
from sklearn.compose import TransformedTargetRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

def rmsle(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    y_pred = np.clip(y_pred, 0, None)
    return np.sqrt(mean_squared_log_error(y_true, y_pred))
rmsle_scorer = make_scorer(rmsle, greater_is_better=False)
warnings.filterwarnings('ignore')

cv = KFold(n_splits=5, shuffle=True, random_state=42)

num_cols = X.select_dtypes(include = np.number).columns.tolist()
cat_cols = X.select_dtypes(include = ['object', 'category', 'str']).columns.tolist()
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'median', keep_empty_features = True))
                        ])
cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy = 'most_frequent', keep_empty_features = True)),
    ('encoder', OneHotEncoder(handle_unknown = 'ignore', sparse_output = False))
                        ])
preprocessor = ColumnTransformer([
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])
def create_pipe(model, scale = False):
    steps = [('preprocessor', preprocessor)]
    if scale:
        steps.append(('scaler', StandardScaler()))
    steps.append(('model', model))
    return Pipeline(steps)

experiments = {
    'LinearRegression': (
        create_pipe(LinearRegression(), scale=True),
        {}
    ),

    'Ridge': (
        create_pipe(Ridge(), scale=True),
        {
            'model__alpha': [0.01, 0.1, 1, 10, 100]
        }
    ),

    'Lasso': (
        create_pipe(Lasso(max_iter=20000), scale=True),
        {
            'model__alpha': [0.0001, 0.001, 0.01, 0.1, 1]
        }
    ),

    'ElasticNet': (
        create_pipe(ElasticNet(max_iter=20000), scale=True),
        {
            'model__alpha': [0.0001, 0.001, 0.01, 0.1],
            'model__l1_ratio': [0.2, 0.5, 0.8]
        }
    ),

    'Huber': (
        create_pipe(HuberRegressor(max_iter=1000), scale=True),
        {
            'model__epsilon': [1.2, 1.35, 1.5],
            'model__alpha': [0.0001, 0.001, 0.01]
        }
    ),
    'KNN': (
        create_pipe(KNeighborsRegressor(), scale = True),
        {
            'model__n_neighbors': [5, 7, 9],
            'model__weights': ['uniform', 'distance']
        }
    ),
    'RandomForest': (
        create_pipe(RandomForestRegressor(random_state=42, n_jobs=-1)),
        {
            'model__n_estimators': [300, 500],
            'model__max_depth': [3, 5, 7],
            'model__min_samples_leaf': [1, 2, 4],
            'model__max_features': [0.7, 1.0]
        }
    ),
    'XGBoost': (
        create_pipe(XGBRegressor(random_state=42, objective = 'reg:squarederror', eval_metric='rmse', n_jobs=1)),
        {
            'model__n_estimators': [300, 500],
            'model__learning_rate': [0.03, 0.05, 0.1],
            'model__max_depth': [2, 3, 4],
            'model__subsample': [0.8, 1.0],
            'model__colsample_bytree': [0.8, 1.0]
        }
    ),
    'LightGBM': (
        create_pipe(LGBMRegressor(random_state=42, verbose=-1, n_jobs=1)),
        {
            'model__n_estimators': [300, 500],
            'model__learning_rate': [0.03, 0.05, 0.1],
            'model__max_depth': [-1, 3, 5],
            'model__num_leaves': [7, 15, 31]
        }
    ),
    'CatBoost': (
        create_pipe(CatBoostRegressor(random_state=42, verbose=0, thread_count=1)),
        {
            'model__iterations': [300, 500],
            'model__learning_rate': [0.03, 0.05, 0.1],
            'model__depth': [4, 6, 8]
        }
    )
}

best_estimators = {}
res_summary = []

for name, (pipeline, param_grid) in experiments.items():
    model = TransformedTargetRegressor(
        regressor = pipeline,
        func = np.log1p,
        inverse_func = np.expm1
    )
    
    grid = GridSearchCV(
        estimator = model,
        param_grid = {
            f'regressor__{key}': value
            for key, value in param_grid.items()
        },
        cv=cv,
        scoring= rmsle_scorer,
        n_jobs = -1,
        verbose = 1
    )
    grid.fit(X, y)
    best_index = grid.best_index_
    best_estimators[name] = grid.best_estimator_
    res_summary.append({
        'Model': name,
        'Best RMSLE': -grid.best_score_,
        'Std': grid.cv_results_['std_test_score'][best_index],
        'Best Params': grid.best_params_
    })
    print(f"[{name}] Best RMSLE: {-grid.best_score_:.4f}")

results_dt = pd.DataFrame(res_summary).sort_values(by='Best RMSLE').reset_index(drop = True)
pd.set_option('display.max_colwidth', None)
display(results_dt)

```
