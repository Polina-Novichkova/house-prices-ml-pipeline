```python
#Пропуски

train_miss = pd.DataFrame({
    "miss_count": train_df.isnull().sum(),
    "miss_percent": train_df.isnull().mean() * 100
})

train_miss = train_miss[train_miss["miss_count"] > 0].sort_values("miss_percent", ascending = False)

test_miss = pd.DataFrame({
    "miss_count": test_df.isnull().sum(),
    "miss_percent": test_df.isnull().mean() * 100
})

test_miss = test_miss[test_miss["miss_count"] > 0].sort_values("miss_percent", ascending = False)

print("Пропуски в train:")
display(train_miss)

print("Пропуски в test:")
display(test_miss)
```

```python
#Проверка наличие гаража, подвала, камина и облицовки

check_cols = [
    "GarageType",
    "GarageCars",
    "GarageArea",
    "GarageYrBlt",
    "TotalBsmtSF",
    "BsmtQual",
    "Fireplaces",
    "FireplaceQu",
    "MasVnrType",
    "MasVnrArea",
    "Fence"
]

display(train_df[check_cols].head(10))
```

```python
#Сколько нулевых значений в важных числовых признаках

zero_cols = [
    "GarageCars",
    "GarageArea",
    "TotalBsmtSF",
    "Fireplaces",
    "MasVnrArea"
]

display((train_df[zero_cols] == 0).sum().to_frame("zero_count"))
```

```python
#Проверяем, согласованы ли пропуски и нулевые значения

checks = {
    "GarageType пропущен, но GarageCars > 0":
        train_df["GarageType"].isna() & (train_df["GarageCars"] > 0),

    "GarageType пропущен, но GarageArea > 0":
        train_df["GarageType"].isna() & (train_df["GarageArea"] > 0),

    "GarageYrBlt пропущен, но гараж есть":
        train_df["GarageYrBlt"].isna() & (train_df["GarageCars"] > 0),

    "BsmtQual пропущен, но TotalBsmtSF > 0":
        train_df["BsmtQual"].isna() & (train_df["TotalBsmtSF"] > 0),

    "BsmtQual пропущен, а TotalBsmtSF = 0":
        train_df["BsmtQual"].isna() & (train_df["TotalBsmtSF"] == 0),

    "FireplaceQu пропущен, но Fireplaces > 0":
        train_df["FireplaceQu"].isna() & (train_df["Fireplaces"] > 0),

    "MasVnrType пропущен, но MasVnrArea > 0":
        train_df["MasVnrType"].isna() & (train_df["MasVnrArea"] > 0),

    "MasVnrType пропущен, а MasVnrArea = 0":
        train_df["MasVnrType"].isna() & (train_df["MasVnrArea"] == 0),
}

for description, condition in checks.items():
    print(f"{description}: {condition.sum()}")
```

```python
#LotFrontage по районам
lot_frontage_stats = (
    train_df.groupby("Neighborhood")["LotFrontage"]
    .agg(["count", "median", "mean"])
    .sort_values("median")
)
display(lot_frontage_stats)
```

```python
#Пропуски LotFrontage в районах
lot_frontage_miss = (
    train_df.groupby("Neighborhood")["LotFrontage"]
    .apply(lambda x: x.isna().sum())
    .sort_values(ascending = False)
)
display(lot_frontage_miss[lot_frontage_miss > 0])
```

```python
#Количество домов и пропусков LotFrontage по районам

lot_frontage_by_neighborhood = train_df.groupby("Neighborhood").agg(
    total_houses=("LotFrontage", "size"),
    known_values=("LotFrontage", "count"),
    missing_values=("LotFrontage", lambda x: x.isna().sum()),
    median=("LotFrontage", "median")
)

lot_frontage_by_neighborhood["missing_percent"] = (
    lot_frontage_by_neighborhood["missing_values"]
    / lot_frontage_by_neighborhood["total_houses"]
    * 100
)

display(
    lot_frontage_by_neighborhood
    .sort_values("missing_percent", ascending=False)
)
```

```python
#Корреляция без нулевых значений
import pandas as pd
corrs = {}
for col in train_df_num.columns:
    if col != 'SalePrice':
        non_zero = train_df_num[train_df_num[col] > 0]
        if len(non_zero) > 0:
            corrs[col] = non_zero[col].corr(non_zero['SalePrice'])

all_correlations = pd.Series(corrs)

golden_features_list = all_correlations[all_correlations.abs() >= 0.5].index.tolist()

print(f"There is {len(golden_features_list)} strongly correlated values with SalePrice:\n{golden_features_list}")
```
